# Functions to perform analysis

```python
def numerical_analysis(dataframe, column_name, cat_col=None, bins="auto"):
    # create the figure
    fig = plt.figure(figsize=(15,10))
    # generate the layout
    grid = GridSpec(nrows=2, ncols=2, figure=fig)
    # set subplots
    ax1 = fig.add_subplot(grid[0, 0])
    ax2 = fig.add_subplot(grid[0, 1])
    ax3 = fig.add_subplot(grid[1, :])
    # plot the kdeplot
    sns.kdeplot(data=dataframe, x=column_name,hue=cat_col, ax=ax1)
    # plot the boxplot
    sns.boxplot(data=dataframe, x=column_name,hue=cat_col, ax=ax2)
    # plot the histogram
    sns.histplot(data=dataframe, x=column_name,bins=bins,hue=cat_col,kde=True, ax=ax3)
    plt.tight_layout()
    plt.show()


def numerical_categorical_analysis(dataframe, cat_column_1, num_column):
    fig, (ax1, ax2) = plt.subplots(2, 2, figsize=(15,7.5))
    # plot the barplot
    sns.barplot(data=dataframe, x=cat_column_1, y=num_column, ax=ax1[0])
    # plot the boxplot
    sns.boxplot(data=dataframe, x=cat_column_1, y=num_column, ax=ax1[1])
    # plot violin plot
    sns.violinplot(data=dataframe, x=cat_column_1, y=num_column, ax=ax2[0])
    # plot strip plot
    sns.stripplot(data=dataframe, x=cat_column_1, y=num_column, ax=ax2[1])
    plt.tight_layout()
    plt.show()


def categorical_analysis(dataframe, column_name):
    # print the values counts of categories
    display(
        pd.DataFrame({
            "Count": (
                dataframe[column_name]
                .value_counts()),
            "Percentage": (
                dataframe[column_name]
                .value_counts(normalize=True)
                .mul(100)
                .round(2)
                .astype("str")
                .add("%")
                )
        })
    )
    print("*" * 50)
    # get unique categories
    unique_categories = dataframe[column_name].unique().tolist()
    number_of_categories = dataframe[column_name].nunique()
    print(f"The unique categories in {column_name} column are {unique_categories}")
    print("*" * 50)
    print(f"The number of categories in {column_name} column are {number_of_categories}")
    # plot countplot
    sns.countplot(data=dataframe, x=column_name)
    plt.xticks(rotation=45)
    plt.show()


def multivariate_analysis(dataframe, num_column, cat_column_1, cat_column_2):
    fig, (ax1, ax2) = plt.subplots(2, 2, figsize=(15,7.5))
    # plot the barplot
    sns.barplot(data=dataframe, x=cat_column_1,
                y=num_column,hue=cat_column_2, ax=ax1[0])
    # plot the boxplot
    sns.boxplot(data=dataframe, x=cat_column_1,
                y=num_column,hue=cat_column_2, gap=0.1, ax=ax1[1])
    # plot violin plot
    sns.violinplot(data=dataframe, x=cat_column_1, gap=0.1,
                   y=num_column,hue=cat_column_2, ax=ax2[0])
    # plot strip plot
    sns.stripplot(data=dataframe, x=cat_column_1,
                  y=num_column,hue=cat_column_2,dodge=True,ax=ax2[1])
    plt.tight_layout()
    plt.show()

def chi_2_test(dataframe, col1, col2, alpha= 0.05):
    data = (
        dataframe.loc[:, [col1, col2]]
        .dropna()
    )
    # create contingency table
    contingency_table = pd.crosstab(data[col1], data[col2])
    # perform chi-squared test
    _, p_val, _, _ = chi2_contingency(contingency_table)
    print(p_val)
    if p_val <= alpha:
        print(f"Reject the null hypothesis. There is a significant association between {col1} and {col2}.")
    else:
        print(f"Fail to reject the null hypothesis. There is no significant association between {col1} and {col2}.")

def anova_test(dataframe, num_col, cat_col, alpha=0.05):
    data = (
        dataframe.loc[:, [num_col, cat_col]]
        .dropna()
    )
    cat_group = data.groupby(cat_col)
    groups = [group[num_col].values for _, group in cat_group]
    f_stat, p_val = f_oneway(*groups)
    print(p_val)
    if p_val <= alpha:
        print(f"Reject the null hypothesis. There is a significant relationship between {num_col} and {cat_col}.")
    else:
        print(f"Fail to reject the null hypothesis. There is no significant relationship between {num_col} and {cat_col}.")

def test_for_normality(dataframe, column_name, alpha=0.05):
    data = dataframe[column_name]
    print("Jarque Bera Test for Normality")
    _, p_val = jarque_bera(data)
    print(p_val)
    if p_val <= alpha:
        print(f"Reject the null hypothesis. The data is not normally distributed.")
    else:
        print(f"Fail to reject the null hypothesis. The data is normally distributed.",end="\n\n")
```

# Hypothesis Testing

- Hypothesis testing = using data to check whether a difference/relationship is real or just happened by chance.
- Example: "Does traffic density really affect delivery time?"

- Every hypothesis test basically starts with:
- H₀ (Null hypothesis) → "There is NO difference/relationship."
- H₁ (Alternative hypothesis) → "There IS a difference/relationship."

- Example:Does traffic affect delivery time?
    - H₀: Traffic does not affect delivery time.
    - H₁: Traffic does affect delivery time.

- Then we calculate a p-value.
- p-value = tells us how likely our result could have happened just by chance if H₀ were true.
- Usually:
    - p-value < 0.05 → Reject H₀ → Something significant is happening
    - p-value ≥ 0.05 → Fail to reject H₀ → Not enough evidence of a difference/relationship


---
- 🎯 TESTS FOR CATEGORICAL DATA (When you have categories/groups)

| Test Name | When to Use | 1-Line Explanation |
|-----------|-------------|-------------------|
| **Chi-Square Test of Independence** | 2 categorical variables | Checks if two categories (like "weather" and "traffic") are related |
| **Chi-Square Goodness of Fit** | 1 categorical variable | Checks if your category counts match expected percentages |
| **Fisher's Exact Test** | 2 categorical variables (small sample) | Like Chi-Square but for small datasets |
| **McNemar's Test** | Paired categorical data | Checks if things changed before vs after (like before/after training) |

---

- 📈 TESTS FOR NUMERICAL DATA (When you have numbers)

- **Comparing 1 Group**

| Test Name | When to Use | 1-Line Explanation |
|-----------|-------------|-------------------|
| **One-Sample t-test** | Compare mean to a known value | Checks if your average is different from a specific number |
| **Wilcoxon Signed-Rank Test** | Compare median to a known value | Like t-test but when data is not normal |
| **Jarque-Bera Test** | Check if data is normal | Checks if your data follows a bell curve |
| **Shapiro-Wilk Test** | Check if data is normal | Another way to check for normal distribution |
| **Kolmogorov-Smirnov Test** | Compare to any distribution | Checks if your data matches any theoretical distribution |

---

-# **Comparing 2 Groups**

| Test Name | When to Use | 1-Line Explanation |
|-----------|-------------|-------------------|
| **Independent t-test** | 2 independent groups | Checks if averages differ between two groups (like men vs women) |
| **Paired t-test** | 2 related groups | Checks if averages differ before vs after (like before/after treatment) |
| **Mann-Whitney U Test** | 2 groups, not normal | Like t-test but when data is not normal |
| **Wilcoxon Signed-Rank Test** | 2 related groups, not normal | Like paired t-test but when data is not normal |

---

-# **Comparing 3+ Groups**

| Test Name | When to Use | 1-Line Explanation |
|-----------|-------------|-------------------|
| **One-Way ANOVA** | 1 category, 1 number | Checks if averages differ across 3+ groups |
| **Two-Way ANOVA** | 2 categories, 1 number | Checks if averages differ across 2 categories at once |
| **Repeated Measures ANOVA** | Repeated measurements on same group | Checks if averages change over time for same people |
| **Kruskal-Wallis Test** | 3+ groups, not normal | Like ANOVA but when data is not normal |
| **Friedman Test** | Repeated measures, not normal | Like Repeated Measures ANOVA but when data is not normal |

---

- 🔗 TESTS FOR RELATIONSHIPS (How variables connect)

| Test Name | When to Use | 1-Line Explanation |
|-----------|-------------|-------------------|
| **Pearson Correlation** | 2 numerical variables | Checks if two numbers move together (like height and weight) |
| **Spearman Correlation** | 2 numerical variables (not normal) | Like Pearson but for non-normal data |
| **Kendall's Tau** | 2 numerical variables | Another way to check if two numbers are related |
| **Point-Biserial Correlation** | 1 number, 1 category with 2 groups | Checks if a number relates to a yes/no category |
| **Multiple Regression** | Many variables | Checks how many factors together affect one number |

---

- 📉 TESTS FOR DIFFERENCES OVER TIME

| Test Name | When to Use | 1-Line Explanation |
|-----------|-------------|-------------------|
| **Paired t-test** | Before vs After | Checks if something changed after an event |
| **Repeated Measures ANOVA** | Multiple time points | Checks if values change across many time points |
| **Friedman Test** | Multiple time points, not normal | Like Repeated Measures ANOVA but for non-normal data |

---

- 🎲 OTHER USEFUL TESTS

| Test Name | When to Use | 1-Line Explanation |

| **Bartlett's Test** | Before ANOVA | Another way to check if groups have similar spread |

- 🎯 QUICK REFERENCE: WHICH TEST TO USE?

| Your Question | Test to Use |
|---------------|-------------|
| "Are two categories related?" | Chi-Square Test |
| "Is my data normal?" | Jarque-Bera or Shapiro-Wilk |
| "Does group A differ from group B on a number?" | t-test (or Mann-Whitney if not normal) |
| "Do 3+ groups differ on a number?" | ANOVA (or Kruskal-Wallis if not normal) |
| "Are two numbers related?" | Pearson Correlation (or Spearman if not normal) |
| "Did something change before vs after?" | Paired t-test (or Wilcoxon if not normal) |
| "Did something change over many time points?" | Repeated Measures ANOVA (or Friedman if not normal) |

---

- 📝 REMEMBER THIS RULE OF THUMB:

- **Data normal?** → Use **parametric tests** (t-test, ANOVA, Pearson)
- **Data NOT normal?** → Use **non-parametric tests** (Mann-Whitney, Kruskal-Wallis, Spearman)
- **Both variables are categories?** → Use **Chi-Square**
- **Both variables are numbers?** → Use **Correlation**

---

**In 1 line:** Choose your test based on: **1) What type of data you have, 2) How many groups, and 3) Whether your data is normal!**
